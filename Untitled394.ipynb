```python
# ============================================================
# INTELLIGENT SCHOLARSHIP ELIGIBILITY PREDICTOR
# FINAL – FULLY CORRECT PIPELINE
# ============================================================

import os
import json
import yaml
import pickle
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, MinMaxScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, roc_auc_score

import lightgbm as lgb
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# ============================================================
# PATHS
# ============================================================

BASE_DIR = r"C:\Users\NXTWAVE\Downloads\Intelligent Scholarship Eligibility Predictor"
DATA_DIR = os.path.join(BASE_DIR, "archive")
os.makedirs(BASE_DIR, exist_ok=True)

MAT_PATH = os.path.join(DATA_DIR, "student-mat.csv")
POR_PATH = os.path.join(DATA_DIR, "student-por.csv")

# ============================================================
# ROBUST CSV LOADER (HANDLES BROKEN CSV)
# ============================================================

def load_student_csv(path):
    with open(path, "r", encoding="utf-8") as f:
        lines = f.read().strip().splitlines()

    header_line = lines[0]
    delimiter = ";" if header_line.count(";") > header_line.count(",") else ","

    header = header_line.split(delimiter)
    data = [row.split(delimiter) for row in lines[1:]]

    df = pd.DataFrame(data, columns=header)
    return df

df_mat = load_student_csv(MAT_PATH)
df_por = load_student_csv(POR_PATH)

print("MAT shape:", df_mat.shape)
print("POR shape:", df_por.shape)
print("Columns:", df_mat.columns.tolist())

# ============================================================
# NUMERIC CONVERSION
# ============================================================

num_cols = [
    'age','Medu','Fedu','traveltime','studytime','failures',
    'famrel','freetime','goout','Dalc','Walc','health',
    'absences','G1','G2','G3'
]

for col in num_cols:
    df_mat[col] = pd.to_numeric(df_mat[col], errors="coerce")
    df_por[col] = pd.to_numeric(df_por[col], errors="coerce")

# ============================================================
# MERGE (EQUIVALENT TO student-merge.R)
# ============================================================

merge_cols = [c for c in df_mat.columns if c not in ['G1','G2','G3']]

df = pd.merge(
    df_mat,
    df_por,
    on=merge_cols,
    suffixes=("_mat", "_por")
)

print("Merged shape:", df.shape)

# ============================================================
# TARGET CREATION (FIXED LOGIC)
# ============================================================

df["avg_grade"] = (df["G3_mat"] + df["G3_por"]) / 2
df["income_proxy"] = df["Medu"] + df["Fedu"]

# absences is a merge key → no suffix
df["attendance"] = df["absences"]

df["eligible"] = (
    (df["avg_grade"] >= 12) &
    (df["income_proxy"] <= 4) &
    (df["attendance"] <= 10)
).astype(int)

# ============================================================
# FEATURES / LABEL
# ============================================================

drop_cols = [
    "eligible", "avg_grade",
    "G1_mat","G2_mat","G3_mat",
    "G1_por","G2_por","G3_por"
]

X = df.drop(columns=drop_cols)
y = df["eligible"]

# Encode categorical features
for col in X.select_dtypes(include="object").columns:
    X[col] = LabelEncoder().fit_transform(X[col])

# ============================================================
# SCALING
# ============================================================

scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)

pickle.dump(scaler, open(os.path.join(BASE_DIR, "scaler.pkl"), "wb"))

# ============================================================
# TRAIN / TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y,
    test_size=0.2,
    random_state=42,
    stratify=y
)

# ============================================================
# LOGISTIC REGRESSION
# ============================================================

log_model = LogisticRegression(max_iter=3000)
log_model.fit(X_train, y_train)

log_pred = log_model.predict(X_test)
log_prob = log_model.predict_proba(X_test)[:, 1]

# ============================================================
# LIGHTGBM
# ============================================================

lgb_model = lgb.LGBMClassifier(
    n_estimators=300,
    learning_rate=0.05,
    max_depth=6
)
lgb_model.fit(X_train, y_train)

lgb_pred = lgb_model.predict(X_test)
lgb_prob = lgb_model.predict_proba(X_test)[:, 1]

# ============================================================
# METRICS
# ============================================================

results = {
    "logistic_accuracy": float(accuracy_score(y_test, log_pred)),
    "logistic_auc": float(roc_auc_score(y_test, log_prob)),
    "lightgbm_accuracy": float(accuracy_score(y_test, lgb_pred)),
    "lightgbm_auc": float(roc_auc_score(y_test, lgb_prob))
}

json.dump(results, open(os.path.join(BASE_DIR, "results.json"), "w"), indent=4)

# ============================================================
# SAVE MODELS (.pkl)
# ============================================================

pickle.dump(log_model, open(os.path.join(BASE_DIR, "logistic_model.pkl"), "wb"))
pickle.dump(lgb_model, open(os.path.join(BASE_DIR, "lightgbm_model.pkl"), "wb"))

# ============================================================
# YAML CONFIG
# ============================================================

yaml.dump(
    {
        "models": ["LogisticRegression", "LightGBM", "NeuralNetwork"],
        "target": "eligible",
        "num_features": X.shape[1],
        "merged_rows": int(df.shape[0])
    },
    open(os.path.join(BASE_DIR, "config.yaml"), "w")
)

# ============================================================
# NEURAL NETWORK (.h5)
# ============================================================

nn = Sequential([
    Dense(64, activation="relu", input_shape=(X_train.shape[1],)),
    Dense(32, activation="relu"),
    Dense(1, activation="sigmoid")
])

nn.compile(optimizer="adam", loss="binary_crossentropy", metrics=["accuracy"])
nn.fit(X_train, y_train, epochs=20, batch_size=16, verbose=0)
nn.save(os.path.join(BASE_DIR, "eligibility_model.h5"))

# ============================================================
# SAVE PREDICTIONS (.json)
# ============================================================

json.dump(
    {
        "true_labels": y_test.tolist(),
        "logistic_probability": log_prob.tolist(),
        "lightgbm_probability": lgb_prob.tolist()
    },
    open(os.path.join(BASE_DIR, "predictions.json"), "w"),
    indent=4
)

print("✅ PIPELINE COMPLETED SUCCESSFULLY")
print("📁 All files saved in:", BASE_DIR)
```
