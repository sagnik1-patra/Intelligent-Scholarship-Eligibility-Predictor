```python
import os
import json
import pickle
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.metrics import (
    accuracy_score,
    roc_auc_score,
    roc_curve,
    confusion_matrix
)

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, MinMaxScaler
from sklearn.linear_model import LogisticRegression
import lightgbm as lgb

# ============================================================
# PATHS
# ============================================================

BASE_DIR = r"C:\Users\NXTWAVE\Downloads\Intelligent Scholarship Eligibility Predictor"
DATA_DIR = os.path.join(BASE_DIR, "archive")

MAT_PATH = os.path.join(DATA_DIR, "student-mat.csv")
POR_PATH = os.path.join(DATA_DIR, "student-por.csv")

# ============================================================
# ROBUST CSV LOADER
# ============================================================

def load_student_csv(path):
    with open(path, "r", encoding="utf-8") as f:
        lines = f.read().strip().splitlines()

    delimiter = ";" if lines[0].count(";") > lines[0].count(",") else ","
    header = lines[0].split(delimiter)
    data = [row.split(delimiter) for row in lines[1:]]

    return pd.DataFrame(data, columns=header)

df_mat = load_student_csv(MAT_PATH)
df_por = load_student_csv(POR_PATH)

# Numeric conversion
num_cols = [
    'age','Medu','Fedu','traveltime','studytime','failures',
    'famrel','freetime','goout','Dalc','Walc','health',
    'absences','G1','G2','G3'
]

for col in num_cols:
    df_mat[col] = pd.to_numeric(df_mat[col], errors="coerce")
    df_por[col] = pd.to_numeric(df_por[col], errors="coerce")

# ============================================================
# MERGE
# ============================================================

merge_cols = [c for c in df_mat.columns if c not in ['G1','G2','G3']]

df = pd.merge(
    df_mat,
    df_por,
    on=merge_cols,
    suffixes=("_mat", "_por")
)

# ============================================================
# TARGET
# ============================================================

df["avg_grade"] = (df["G3_mat"] + df["G3_por"]) / 2
df["income_proxy"] = df["Medu"] + df["Fedu"]
df["attendance"] = df["absences"]

df["eligible"] = (
    (df["avg_grade"] >= 12) &
    (df["income_proxy"] <= 4) &
    (df["attendance"] <= 10)
).astype(int)

# ============================================================
# FEATURES
# ============================================================

drop_cols = [
    "eligible", "avg_grade",
    "G1_mat","G2_mat","G3_mat",
    "G1_por","G2_por","G3_por"
]

X = df.drop(columns=drop_cols)
y = df["eligible"]

for col in X.select_dtypes(include="object").columns:
    X[col] = LabelEncoder().fit_transform(X[col])

scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)

# ============================================================
# TRAIN / TEST
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y,
    test_size=0.2,
    random_state=42,
    stratify=y
)

# ============================================================
# MODELS
# ============================================================

log_model = LogisticRegression(max_iter=3000)
log_model.fit(X_train, y_train)

lgb_model = lgb.LGBMClassifier(
    n_estimators=300,
    learning_rate=0.05,
    max_depth=6
)
lgb_model.fit(X_train, y_train)

# Predictions
log_pred = log_model.predict(X_test)
log_prob = log_model.predict_proba(X_test)[:, 1]

lgb_pred = lgb_model.predict(X_test)
lgb_prob = lgb_model.predict_proba(X_test)[:, 1]

# ============================================================
# METRICS
# ============================================================

log_acc = accuracy_score(y_test, log_pred)
lgb_acc = accuracy_score(y_test, lgb_pred)

log_auc = roc_auc_score(y_test, log_prob)
lgb_auc = roc_auc_score(y_test, lgb_prob)

# ============================================================
# 1️⃣ ACCURACY COMPARISON GRAPH
# ============================================================

plt.figure()
plt.bar(["Logistic", "LightGBM"], [log_acc, lgb_acc])
plt.title("Model Accuracy Comparison")
plt.ylabel("Accuracy")
plt.savefig(os.path.join(BASE_DIR, "accuracy_comparison.png"))
plt.show()

# ============================================================
# 2️⃣ ROC CURVE (RESULT GRAPH)
# ============================================================

fpr_log, tpr_log, _ = roc_curve(y_test, log_prob)
fpr_lgb, tpr_lgb, _ = roc_curve(y_test, lgb_prob)

plt.figure()
plt.plot(fpr_log, tpr_log, label="Logistic Regression")
plt.plot(fpr_lgb, tpr_lgb, label="LightGBM")
plt.plot([0,1], [0,1], linestyle="--")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve")
plt.legend()
plt.savefig(os.path.join(BASE_DIR, "roc_curve.png"))
plt.show()

# ============================================================
# 3️⃣ CONFUSION MATRIX HEATMAP
# ============================================================

cm = confusion_matrix(y_test, lgb_pred)

plt.figure()
sns.heatmap(cm, annot=True, fmt="d")
plt.title("Confusion Matrix (LightGBM)")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.savefig(os.path.join(BASE_DIR, "confusion_matrix_heatmap.png"))
plt.show()

# ============================================================
# 4️⃣ PREDICTION COMPARISON GRAPH
# ============================================================

plt.figure()
plt.plot(log_prob[:30], label="Logistic Prob")
plt.plot(lgb_prob[:30], label="LightGBM Prob")
plt.title("Prediction Probability Comparison (Sample)")
plt.ylabel("Probability")
plt.xlabel("Sample Index")
plt.legend()
plt.savefig(os.path.join(BASE_DIR, "prediction_comparison.png"))
plt.show()

# ============================================================
# 5️⃣ RESULTS CSV
# ============================================================

results_df = pd.DataFrame({
    "Model": ["Logistic Regression", "LightGBM"],
    "Accuracy": [log_acc, lgb_acc],
    "AUC": [log_auc, lgb_auc]
})

results_df.to_csv(
    os.path.join(BASE_DIR, "results_summary.csv"),
    index=False
)

# ============================================================
# 6️⃣ PREDICTIONS JSON
# ============================================================

predictions = {
    "true_labels": y_test.tolist(),
    "logistic_probability": log_prob.tolist(),
    "lightgbm_probability": lgb_prob.tolist()
}

with open(os.path.join(BASE_DIR, "predictions.json"), "w") as f:
    json.dump(predictions, f, indent=4)

print("✅ ALL GRAPHS, CSV, AND JSON SAVED SUCCESSFULLY")
```
