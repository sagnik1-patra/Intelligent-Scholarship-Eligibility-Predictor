```python
# ============================================================
# HYBRID AIS + CSA SCHOLARSHIP ELIGIBILITY PREDICTOR
# ============================================================

import os
import json
import pickle
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, MinMaxScaler
from sklearn.metrics import accuracy_score, roc_auc_score, roc_curve, confusion_matrix
from sklearn.linear_model import LogisticRegression
import lightgbm as lgb

# ============================================================
# PATHS
# ============================================================

BASE_DIR = r"C:\Users\NXTWAVE\Downloads\Intelligent Scholarship Eligibility Predictor"
DATA_DIR = os.path.join(BASE_DIR, "archive")

MAT_PATH = os.path.join(DATA_DIR, "student-mat.csv")
POR_PATH = os.path.join(DATA_DIR, "student-por.csv")

# ============================================================
# ROBUST CSV LOADER
# ============================================================

def load_student_csv(path):
    with open(path, "r", encoding="utf-8") as f:
        lines = f.read().strip().splitlines()
    delimiter = ";" if lines[0].count(";") > lines[0].count(",") else ","
    header = lines[0].split(delimiter)
    data = [row.split(delimiter) for row in lines[1:]]
    return pd.DataFrame(data, columns=header)

df_mat = load_student_csv(MAT_PATH)
df_por = load_student_csv(POR_PATH)

# Numeric conversion
num_cols = [
    'age','Medu','Fedu','traveltime','studytime','failures',
    'famrel','freetime','goout','Dalc','Walc','health',
    'absences','G1','G2','G3'
]

for col in num_cols:
    df_mat[col] = pd.to_numeric(df_mat[col], errors="coerce")
    df_por[col] = pd.to_numeric(df_por[col], errors="coerce")

# ============================================================
# MERGE (R-script equivalent)
# ============================================================

merge_cols = [c for c in df_mat.columns if c not in ['G1','G2','G3']]
df = pd.merge(df_mat, df_por, on=merge_cols, suffixes=("_mat", "_por"))

# ============================================================
# TARGET
# ============================================================

df["avg_grade"] = (df["G3_mat"] + df["G3_por"]) / 2
df["income_proxy"] = df["Medu"] + df["Fedu"]
df["attendance"] = df["absences"]

df["eligible"] = (
    (df["avg_grade"] >= 12) &
    (df["income_proxy"] <= 4) &
    (df["attendance"] <= 10)
).astype(int)

# ============================================================
# FEATURES
# ============================================================

drop_cols = [
    "eligible","avg_grade",
    "G1_mat","G2_mat","G3_mat",
    "G1_por","G2_por","G3_por"
]

X = df.drop(columns=drop_cols)
y = df["eligible"]

for col in X.select_dtypes(include="object").columns:
    X[col] = LabelEncoder().fit_transform(X[col])

scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)

pickle.dump(scaler, open(os.path.join(BASE_DIR, "hybrid_scaler.pkl"), "wb"))

# ============================================================
# TRAIN / TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y,
    test_size=0.2,
    random_state=42,
    stratify=y
)

# ============================================================
# AIS FEATURE SELECTION (SIMPLIFIED)
# ============================================================

def ais_feature_selection(X, y, n_features=10):
    scores = []
    for i in range(X.shape[1]):
        model = LogisticRegression(max_iter=2000)
        model.fit(X[:, [i]], y)
        score = model.score(X[:, [i]], y)
        scores.append(score)
    top_idx = np.argsort(scores)[-n_features:]
    return top_idx

selected_features = ais_feature_selection(X_train, y_train, n_features=12)

X_train_ais = X_train[:, selected_features]
X_test_ais = X_test[:, selected_features]

# ============================================================
# CSA HYPERPARAMETER OPTIMIZATION
# ============================================================

def csa_optimize(X, y, iters=10):
    best_score = 0
    best_params = None

    for _ in range(iters):
        params = {
            "n_estimators": random.randint(100, 400),
            "learning_rate": random.uniform(0.01, 0.2),
            "max_depth": random.randint(3, 8)
        }

        model = lgb.LGBMClassifier(**params)
        model.fit(X, y)
        score = model.score(X, y)

        if score > best_score:
            best_score = score
            best_params = params

    return best_params

best_params = csa_optimize(X_train_ais, y_train)

# ============================================================
# FINAL HYBRID MODEL
# ============================================================

hybrid_model = lgb.LGBMClassifier(**best_params)
hybrid_model.fit(X_train_ais, y_train)

hybrid_pred = hybrid_model.predict(X_test_ais)
hybrid_prob = hybrid_model.predict_proba(X_test_ais)[:, 1]

# ============================================================
# METRICS
# ============================================================

hybrid_acc = accuracy_score(y_test, hybrid_pred)
hybrid_auc = roc_auc_score(y_test, hybrid_prob)

# ============================================================
# SAVE MODEL
# ============================================================

pickle.dump(
    hybrid_model,
    open(os.path.join(BASE_DIR, "hybrid_lightgbm_model.pkl"), "wb")
)

# ============================================================
# GRAPHS
# ============================================================

# Accuracy graph
plt.figure()
plt.bar(["Hybrid AIS+CSA"], [hybrid_acc])
plt.title("Hybrid AIS + CSA Accuracy")
plt.savefig(os.path.join(BASE_DIR, "hybrid_accuracy_comparison.png"))
plt.show()

# ROC
fpr, tpr, _ = roc_curve(y_test, hybrid_prob)
plt.figure()
plt.plot(fpr, tpr, label=f"AUC = {hybrid_auc:.2f}")
plt.plot([0,1], [0,1], linestyle="--")
plt.legend()
plt.title("Hybrid ROC Curve")
plt.savefig(os.path.join(BASE_DIR, "hybrid_roc_curve.png"))
plt.show()

# Heatmap
cm = confusion_matrix(y_test, hybrid_pred)
plt.figure()
sns.heatmap(cm, annot=True, fmt="d")
plt.title("Hybrid Confusion Matrix")
plt.savefig(os.path.join(BASE_DIR, "hybrid_confusion_matrix_heatmap.png"))
plt.show()

# Prediction graph
plt.figure()
plt.plot(hybrid_prob[:30])
plt.title("Hybrid Prediction Probabilities")
plt.savefig(os.path.join(BASE_DIR, "hybrid_prediction_comparison.png"))
plt.show()

# ============================================================
# SAVE CSV + JSON
# ============================================================

pd.DataFrame({
    "Model": ["Hybrid AIS+CSA"],
    "Accuracy": [hybrid_acc],
    "AUC": [hybrid_auc]
}).to_csv(
    os.path.join(BASE_DIR, "hybrid_results_summary.csv"),
    index=False
)

json.dump(
    {
        "true_labels": y_test.tolist(),
        "hybrid_probability": hybrid_prob.tolist()
    },
    open(os.path.join(BASE_DIR, "hybrid_predictions.json"), "w"),
    indent=4
)

print("✅ HYBRID AIS + CSA RESULTS SAVED SUCCESSFULLY")
```
